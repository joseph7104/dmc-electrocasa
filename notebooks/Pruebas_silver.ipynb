# LECTURA DE TABLAS GENERADAS

### CATÁLOGO

```python
%sql
-- Ejemplo de perfilado rápido en ventas
SELECT
  COUNT(*) AS total_filas,
  COUNT(producto_id) AS productos_no_nulas,
  COUNT(DISTINCT producto_id) AS productos_unicas,
  MIN(try_cast(precio_lista AS FLOAT)) AS min_precio,
  MAX(try_cast(precio_lista AS FLOAT)) AS max_precio
FROM electrocasa_dev.silver.catalogo;
```

### DEVOLUCIONES

```python
%sql
-- Ejemplo de perfilado rápido en ventas
SELECT
  COUNT(*) AS total_filas,
  COUNT(devolucion_id) AS productos_no_nulas,
  COUNT(DISTINCT devolucion_id) AS productos_unicas,
  MIN(try_cast(monto_reembolso AS FLOAT)) AS min_monto,
  MAX(try_cast(monto_reembolso AS FLOAT)) AS monto,
  MIN(TRY_CAST(fecha_devolucion AS DATE)) AS min_fecha,
  MAX(TRY_CAST(fecha_devolucion AS DATE)) AS max_fecha
FROM electrocasa_dev.silver.devoluciones;
```

### EMPLEADOS

```python
%sql
-- Ejemplo de perfilado rápido en ventas
SELECT
  COUNT(*) AS total_filas,
  COUNT(id_empleado) AS productos_no_nulas,
  COUNT(DISTINCT id_empleado) AS productos_unicas,
  MIN(try_cast(salario AS FLOAT)) AS min_salario,
  MAX(try_cast(salario AS FLOAT)) AS max_salario,
  MIN(TRY_CAST(fecha_evento AS DATE)) AS min_fecha,
  MAX(TRY_CAST(fecha_evento AS DATE)) AS max_fecha
FROM electrocasa_dev.silver.empleados WHERE __END_AT IS NULL;
```

### VENTAS

```python
%sql
-- Ejemplo de perfilado rápido en ventas
SELECT
  COUNT(*) AS total_filas,
  COUNT(venta_id) AS resenas_resenas,
  COUNT(DISTINCT venta_id) AS resenas_unicas,
  MIN(try_cast(cantidad AS INT)) AS min_cantidad,
  MAX(try_cast(cantidad AS INT)) AS max_cantidad,
  MIN(try_cast(monto_total AS FLOAT)) AS min_monto,
  MAX(try_cast(monto_total AS FLOAT)) AS max_monto,
  MIN(try_cast(fecha_venta AS DATE)) AS min_fecha,
  MAX(try_cast(fecha_venta AS DATE)) AS max_fecha
FROM electrocasa_dev.silver.ventas;
```

### TRACKING
